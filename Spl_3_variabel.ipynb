**Cell 1**

```python
print("===================================")
print("PROGRAM SPL 3 VARIABEL")
print("METODE ELIMINASI GAUSS")
print("===================================")
```

**Cell 2**

```python
A = [
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
]

# Nilai konstanta
B = [10, 11, 12]

print("Matriks Koefisien:")
for baris in A:
    print(baris)

print("Matriks Konstanta:")
print(B)
```

**Cell 3**

```python
M = [A[i] + [B[i]] for i in range(3)]

n = 3
epsilon = 1e-10
kolom_pivot = []
baris_pivot = 0

# Proses eliminasi Gauss
for kolom in range(n):

    pivot = max(
        range(baris_pivot, n),
        key=lambda r: abs(M[r][kolom])
    )

    if abs(M[pivot][kolom]) < epsilon:
        continue

    # Pertukaran baris
    M[baris_pivot], M[pivot] = M[pivot], M[baris_pivot]

    # Proses eliminasi
    for j in range(baris_pivot + 1, n):
        faktor = M[j][kolom] / M[baris_pivot][kolom]

        for k in range(kolom, n + 1):
            M[j][k] -= faktor * M[baris_pivot][k]

    kolom_pivot.append(kolom)
    baris_pivot += 1

    if baris_pivot == n:
        break

print("Matriks setelah eliminasi:")
for baris in M:
    print(baris)
```

**Cell 4**

```python
tidak_ada_solusi = False

for i in range(n):
    if all(abs(M[i][j]) < epsilon for j in range(n)):
        if abs(M[i][n]) >= epsilon:
            tidak_ada_solusi = True

jumlah_pivot = len(kolom_pivot)

if tidak_ada_solusi:
    print("Sistem tidak memiliki solusi.")

elif jumlah_pivot < n:
    print("Sistem memiliki banyak solusi.")

else:
    hasil = [0.0] * n

    # Substitusi mundur
    for i in range(n - 1, -1, -1):
        kolom = kolom_pivot[i]

        jumlah = M[i][n]

        for j in range(kolom + 1, n):
            jumlah -= M[i][j] * hasil[j]

        hasil[kolom] = jumlah / M[i][kolom]

    print("Nilai x =", hasil[0])
    print("Nilai y =", hasil[1])
    print("Nilai z =", hasil[2])
```

**Cell 5**

```python
print("===================================")
print("KESIMPULAN")
print("===================================")

if tidak_ada_solusi:
    print("SPL tidak memiliki solusi.")

elif jumlah_pivot < n:
    print("SPL memiliki banyak solusi.")

else:
    print(f"Nilai x = {hasil[0]:.2f}")
    print(f"Nilai y = {hasil[1]:.2f}")
    print(f"Nilai z = {hasil[2]:.2f}")

    print("SPL berhasil diselesaikan.")
```
