**Cell 2**

```python
# Matriks koefisien x, y, dan z.
# Setiap baris menunjukkan koefisien dari satu persamaan.
A = [
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
]

# Nilai konstanta dari masing-masing persamaan.
B = [10, 11, 12]

# Menampilkan keterangan matriks koefisien.
print("Matriks Koefisien:")

# Melakukan perulangan untuk menampilkan setiap baris matriks A.
for baris in A:
    print(baris)

# Menampilkan keterangan matriks konstanta.
print("Matriks Konstanta:")

# Menampilkan seluruh nilai konstanta yang terdapat pada matriks B.
print(B)
```

**Cell 3**

```python
# Menggabungkan matriks koefisien dan konstanta menjadi matriks augmented.
M = [A[i] + [B[i]] for i in range(3)]

# Menentukan jumlah variabel atau jumlah kolom koefisien.
n = 3

# Menentukan nilai toleransi untuk memeriksa bilangan yang mendekati nol.
epsilon = 1e-10

# Membuat list kosong untuk menyimpan posisi kolom pivot.
kolom_pivot = []

# Menentukan posisi awal baris pivot, dimulai dari baris pertama.
baris_pivot = 0

# Memulai proses eliminasi Gauss untuk setiap kolom.
for kolom in range(n):

    # Mencari baris yang memiliki nilai absolut terbesar pada kolom saat ini.
    pivot = max(
        range(baris_pivot, n),
        key=lambda r: abs(M[r][kolom])
    )

    # Memeriksa apakah nilai pivot mendekati nol.
    # Jika iya, proses dilanjutkan ke kolom berikutnya.
    if abs(M[pivot][kolom]) < epsilon:
        continue

    # Pertukaran baris untuk menempatkan pivot pada posisi yang tepat.
    M[baris_pivot], M[pivot] = M[pivot], M[baris_pivot]

    # Memulai proses eliminasi pada baris di bawah baris pivot.
    for j in range(baris_pivot + 1, n):

        # Menghitung faktor pengali untuk menghilangkan nilai di bawah pivot.
        faktor = M[j][kolom] / M[baris_pivot][kolom]

        # Melakukan operasi pengurangan pada setiap elemen baris.
        for k in range(kolom, n + 1):

            # Mengubah nilai elemen agar elemen di bawah pivot menjadi nol.
            M[j][k] -= faktor * M[baris_pivot][k]

    # Menyimpan posisi kolom yang memiliki pivot.
    kolom_pivot.append(kolom)

    # Berpindah ke baris pivot berikutnya.
    baris_pivot += 1

    # Menghentikan proses jika seluruh baris sudah memiliki pivot.
    if baris_pivot == n:
        break

# Menampilkan keterangan matriks setelah proses eliminasi.
print("Matriks setelah eliminasi:")

# Menampilkan setiap baris matriks hasil eliminasi Gauss.
for baris in M:
    print(baris)
```

**Cell 4**

```python
# Memeriksa konsistensi sistem persamaan linear.
tidak_ada_solusi = False

# Melakukan pemeriksaan terhadap setiap baris matriks hasil eliminasi.
for i in range(n):

    # Memeriksa apakah seluruh koefisien pada suatu baris mendekati nol.
    if all(abs(M[i][j]) < epsilon for j in range(n)):

        # Jika koefisien nol tetapi konstantanya bukan nol,
        # maka sistem persamaan tidak memiliki solusi.
        if abs(M[i][n]) >= epsilon:
            tidak_ada_solusi = True

# Menghitung jumlah pivot yang ditemukan selama eliminasi.
jumlah_pivot = len(kolom_pivot)

# Memeriksa apakah sistem tidak memiliki solusi.
if tidak_ada_solusi:
    print("Sistem tidak memiliki solusi.")

# Memeriksa apakah jumlah pivot kurang dari jumlah variabel.
# Jika benar, sistem memiliki banyak solusi.
elif jumlah_pivot < n:
    print("Sistem memiliki banyak solusi.")

# Jika sistem memiliki solusi tunggal, lanjutkan substitusi mundur.
else:
    # Membuat list untuk menyimpan hasil nilai setiap variabel.
    hasil = [0.0] * n

    # Memulai proses substitusi mundur dari baris terakhir ke baris pertama.
    for i in range(n - 1, -1, -1):

        # Mengambil posisi kolom pivot pada baris yang sedang dihitung.
        kolom = kolom_pivot[i]

        # Mengambil nilai konstanta dari baris matriks tersebut.
        jumlah = M[i][n]

        # Menghitung kontribusi variabel yang sudah diketahui nilainya.
        for j in range(kolom + 1, n):

            # Mengurangi nilai konstanta dengan hasil perkalian koefisien
            # dan nilai variabel yang sudah dihitung sebelumnya.
            jumlah -= M[i][j] * hasil[j]

        # Menghitung nilai variabel dengan membagi jumlah terhadap pivot.
        hasil[kolom] = jumlah / M[i][kolom]

    # Menampilkan nilai variabel x.
    print("Nilai x =", hasil[0])

    # Menampilkan nilai variabel y.
    print("Nilai y =", hasil[1])

    # Menampilkan nilai variabel z.
    print("Nilai z =", hasil[2])
```

**Cell 5**

```python
# Menampilkan garis pembatas sebagai pemisah tampilan.
print("===================================")

# Menampilkan judul bagian kesimpulan.
print("KESIMPULAN")

# Menampilkan garis pembatas di bawah judul.
print("===================================")

# Memeriksa apakah sistem persamaan tidak memiliki solusi.
if tidak_ada_solusi:
    print("SPL tidak memiliki solusi.")

# Memeriksa apakah sistem memiliki banyak solusi.
elif jumlah_pivot < n:
    print("SPL memiliki banyak solusi.")

# Jika sistem memiliki solusi tunggal, tampilkan hasil akhirnya.
else:

    # Menampilkan nilai variabel x dengan dua angka di belakang koma.
    print(f"Nilai x = {hasil[0]:.2f}")

    # Menampilkan nilai variabel y dengan dua angka di belakang koma.
    print(f"Nilai y = {hasil[1]:.2f}")

    # Menampilkan nilai variabel z dengan dua angka di belakang koma.
    print(f"Nilai z = {hasil[2]:.2f}")

    # Menampilkan keterangan bahwa proses penyelesaian berhasil.
    print("SPL berhasil diselesaikan.")
```
